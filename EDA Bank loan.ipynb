```python
# Drop unnessaory and 40% of data is missing column from Application Data dataframe
cols_to_drop = [
    'COMMONAREA_AVG', 'COMMONAREA_MODE', 'COMMONAREA_MEDI',
    'NONLIVINGAPARTMENTS_MEDI', 'NONLIVINGAPARTMENTS_MODE', 'NONLIVINGAPARTMENTS_AVG',
    'FONDKAPREMONT_MODE',
    'LIVINGAPARTMENTS_AVG', 'LIVINGAPARTMENTS_MEDI', 'LIVINGAPARTMENTS_MODE',
    'FLOORSMIN_MODE', 'FLOORSMIN_AVG', 'FLOORSMIN_MEDI',
    'YEARS_BUILD_AVG', 'YEARS_BUILD_MODE', 'YEARS_BUILD_MEDI',
    'OWN_CAR_AGE',
    'LANDAREA_MEDI', 'LANDAREA_AVG', 'LANDAREA_MODE',
    'BASEMENTAREA_MODE', 'BASEMENTAREA_MEDI', 'BASEMENTAREA_AVG',
    'EXT_SOURCE_1',
    'NONLIVINGAREA_MODE', 'NONLIVINGAREA_AVG', 'NONLIVINGAREA_MEDI',
    'ELEVATORS_AVG', 'ELEVATORS_MEDI', 'ELEVATORS_MODE',
    'WALLSMATERIAL_MODE',
    'APARTMENTS_AVG', 'APARTMENTS_MEDI', 'APARTMENTS_MODE',
    'ENTRANCES_MODE', 'ENTRANCES_MEDI', 'ENTRANCES_AVG',
    'LIVINGAREA_AVG', 'LIVINGAREA_MEDI', 'LIVINGAREA_MODE',
    'HOUSETYPE_MODE',
    'FLOORSMAX_MODE', 'FLOORSMAX_AVG', 'FLOORSMAX_MEDI',
    'YEARS_BEGINEXPLUATATION_MODE', 'YEARS_BEGINEXPLUATATION_MEDI', 'YEARS_BEGINEXPLUATATION_AVG',
    'TOTALAREA_MODE', 'EMERGENCYSTATE_MODE','EXT_SOURCE_3','EXT_SOURCE_2','REGION_POPULATION_RELATIVE','REGION_RATING_CLIENT','REGION_RATING_CLIENT_W_CITY'
    ,'WEEKDAY_APPR_PROCESS_START','HOUR_APPR_PROCESS_START','REG_REGION_NOT_LIVE_REGION','REG_REGION_NOT_WORK_REGION','LIVE_REGION_NOT_WORK_REGION','REG_CITY_NOT_LIVE_CITY'
    ,'REG_CITY_NOT_WORK_CITY','LIVE_CITY_NOT_WORK_CITY'
]

df.drop(columns=cols_to_drop, inplace=True)
```

```python
# It’s categorical, and filling with "Unknown" keeps the data usable without deleting rows.
df['OCCUPATION_TYPE'] = df['OCCUPATION_TYPE'].fillna('Unknown')
df['NAME_TYPE_SUITE'] = df['NAME_TYPE_SUITE'].fillna('Unknown')

# Missingness is extremely small; use median to preserve distribution:
df['AMT_GOODS_PRICE'] = df['AMT_GOODS_PRICE'].fillna(df['AMT_GOODS_PRICE'].median())
df['AMT_ANNUITY'] = df['AMT_ANNUITY'].fillna(df['AMT_ANNUITY'].median())
df['CNT_FAM_MEMBERS'] = df['CNT_FAM_MEMBERS'].fillna(df['CNT_FAM_MEMBERS'].median())
df['DAYS_LAST_PHONE_CHANGE'] = df['DAYS_LAST_PHONE_CHANGE'].fillna(df['DAYS_LAST_PHONE_CHANGE'].median())


# Missing values likely mean no record → fill with 0.
social_cols = ['DEF_30_CNT_SOCIAL_CIRCLE', 'OBS_30_CNT_SOCIAL_CIRCLE',
               'DEF_60_CNT_SOCIAL_CIRCLE', 'OBS_60_CNT_SOCIAL_CIRCLE']
df[social_cols] = df[social_cols].fillna(0)

bureau_cols = ['AMT_REQ_CREDIT_BUREAU_HOUR', 'AMT_REQ_CREDIT_BUREAU_DAY',
               'AMT_REQ_CREDIT_BUREAU_WEEK', 'AMT_REQ_CREDIT_BUREAU_MON',
               'AMT_REQ_CREDIT_BUREAU_QRT', 'AMT_REQ_CREDIT_BUREAU_YEAR']
df[bureau_cols] = df[bureau_cols].fillna(0)

# If output is 0, all missing values are handled successfully.
output = df.isnull().sum().sum()
print(f"Missing Values Count : {output}")
```

<strong>Clean categorical column values</strong>

```python
# NAME_TYPE_SUITE
df['NAME_TYPE_SUITE'] = df['NAME_TYPE_SUITE'].replace({
    'Spouse, partner': 'Spouse'
})

# NAME_INCOME_TYPE - Maternity leave is for working professional
df['NAME_INCOME_TYPE'] = df['NAME_INCOME_TYPE'].replace({
    'Maternity leave': 'Working'
})

# NAME_EDUCATION_TYPE
df['NAME_EDUCATION_TYPE'] = df['NAME_EDUCATION_TYPE'].replace({
    'Secondary / secondary special': 'Secondary'
})

# ORGANIZATION_TYPE
# Simplify to keep only the main category (remove text after ":" and map Business Entity)
df['ORGANIZATION_TYPE'] = (
    df['ORGANIZATION_TYPE']
    .str.replace(r':.*', '', regex=True)  # remove text after colon
    .replace({
        'Business Entity Type 1': 'Business',
        'Business Entity Type 2': 'Business',
        'Business Entity Type 3': 'Business'
    })
)
```

<strong>Relation between Categorical Variables and Default (TARGET)</strong>

```python
colsCategorical = [
    'NAME_CONTRACT_TYPE','CODE_GENDER','FLAG_OWN_CAR','FLAG_OWN_REALTY','NAME_TYPE_SUITE',
    'NAME_INCOME_TYPE','NAME_EDUCATION_TYPE','NAME_FAMILY_STATUS','NAME_HOUSING_TYPE',
    'FLAG_MOBIL','FLAG_EMP_PHONE','FLAG_WORK_PHONE','FLAG_CONT_MOBILE','FLAG_PHONE',
    'FLAG_EMAIL','OCCUPATION_TYPE','ORGANIZATION_TYPE'
]

# Calculate the required grid size (e.g., 4 columns, calculated rows)
n_cols = 4
n_rows = (len(colsCategorical) + n_cols - 1) // n_cols

# Create a figure and a set of subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(4 * n_cols, 4 * n_rows))
# Flatten the axes array for easy iteration
axes = axes.flatten()

for i, col in enumerate(colsCategorical):
    # Calculate % of defaulters in each category
    crosstab_df = pd.crosstab(df[col], df['TARGET'], normalize='index')

    # Plot on the current subplot
    sns.heatmap(
        crosstab_df,
        annot=True,
        fmt='.2f',
        cmap='Paired',
        ax=axes[i], # <--- Specify the subplot
        cbar=False # <--- Turn off individual color bars for cleaner look
    )
    axes[i].set_title(f'{col}')
    axes[i].set_ylabel('') # Optional: clean up y-labels

# Remove any unused subplots
for j in range(len(colsCategorical), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.suptitle('Crosstabulation of Categorical Features vs. TARGET (Defaulter %)', y=1.02, fontsize=16)
plot_path = "application_data_corr.png"
plt.savefig(plot_path, bbox_inches='tight')
plt.show()
```

```python
filterdf = df[df['TARGET'] == 1]
colsChagerical = [
    'CODE_GENDER','NAME_INCOME_TYPE','NAME_EDUCATION_TYPE','NAME_FAMILY_STATUS','OCCUPATION_TYPE'
]

for col in colsChagerical:
    # Calculate % of defaulters in each category
    perc_df = (
        df.groupby(col)['TARGET']
        .mean()
        .sort_values(ascending=False)
        .reset_index()
    )
    perc_df['TARGET'] = perc_df['TARGET'] * 100  # convert to %

    # Plot (fixed palette warning)
    plt.figure(figsize=(15, 4))
    
    sns.barplot(
        x=col,
        y='TARGET',
        hue=col,                # assign hue same as x
        data=perc_df,
        palette='viridis',
        legend=False            # remove duplicate legend
    )

    plt.title(f'Percentage of Defaulters by {col}', fontsize=14)
    plt.ylabel('Defaulter Percentage (%)')
    plt.xlabel(col)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plot_path = f"application_data_corr_{col}.png"
    plt.savefig(plot_path, bbox_inches='tight')
    plt.show()
```

### Key Insights from Categorical Features vs TARGET (Defaulters %)

1. **Employment Status:**
   - **Unemployed (36%)** show the **highest default rate** among all categories.
   - **Businessman (0%)** and **Students (0%)** have the **lowest risk** of default.

2. **Education Level:**
   - Clients with **Lower Secondary Education (11%)** have a **higher chance of default**.
   - Those with an **Academic Degree (2%)** are the most reliable.

3. **Gender:**
   - **Males (10%)** default more than **Females (7%)**.

4. **Loan Type:**
   - **Cash Loans (8%)** have a higher default rate than **Revolving Loans (5%)**.

5. **Housing Type:**
   - People living **Rented (12%)** or **With Parents (12%)** show higher defaults.
   - **House/Apartment owners (8%)** are more stable borrowers.

6. **Family Status:**
   - **Single or Civil Marriage clients (10%)** have higher default rates than **Married (8%)**.

7. **Occupation Type:**
   - **Low-skill Laborers (17%)** show the **highest default** among occupations.
   - **Accountants (5%)** and **High-skill tech staff (6%)** are more creditworthy.

8. **General Observation:**
   - Most categorical groups cluster around **7–9% default rate**.
   - Employment,Education, income type, and housing situation are the **strongest predictors** of loan default likelihood.


<strong>Relation between Numerical Variables and Default (TARGET)</strong>

```python
df['AGE'] = abs(df['DAYS_BIRTH'] // 365)
df['EMPLOYMENT_YEARS'] = abs(df['DAYS_EMPLOYED'] // 365)

# High ratio → higher default (Over-borrowing relative to income)
df['CREDIT_INCOME_RATIO'] = df['AMT_CREDIT'] / df['AMT_INCOME_TOTAL']
# High ratio → higher burden (More difficult repayment)
df['ANNUITY_INCOME_RATIO'] = df['AMT_ANNUITY'] / df['AMT_INCOME_TOTAL']

# Create combined columns
df['TOTAL_FLAG_DOCUMENT'] = df.loc[:, 'FLAG_DOCUMENT_2':'FLAG_DOCUMENT_21'].sum(axis=1)
df['TOTAL_AMT_REQ_CREDIT_BUREAU'] = df.loc[:, 'AMT_REQ_CREDIT_BUREAU_HOUR':'AMT_REQ_CREDIT_BUREAU_YEAR'].sum(axis=1)

# Correlation of numerical features with TARGET
colsNumerical = ['AGE','EMPLOYMENT_YEARS','CREDIT_INCOME_RATIO','ANNUITY_INCOME_RATIO',
    'AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE', 'DAYS_EMPLOYED',
    'DAYS_REGISTRATION', 'DAYS_ID_PUBLISH', 'CNT_FAM_MEMBERS',
    'DAYS_LAST_PHONE_CHANGE','TOTAL_FLAG_DOCUMENT','TOTAL_AMT_REQ_CREDIT_BUREAU'
]

corr = df[colsNumerical + ['TARGET']].corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plot_path = f"application_data_corr_num.png"
plt.savefig(plot_path, bbox_inches='tight')
plt.show()


```

### Key Observations

- All correlations with `TARGET` are **very weak (|r| < 0.1)** → no single numeric variable strongly predicts default.  
- **Negative correlations:**  
  - `AGE` (-0.08) and `EMPLOYMENT_YEARS` (-0.05) → Younger or less experienced clients tend to default slightly more.  
- **Positive correlations:**  
  - `DAYS_LAST_PHONE_CHANGE` (0.06) and `DAYS_ID_PUBLISH` (0.05) → More recent updates or changes in personal info slightly increase default risk.  
- **New ratio features (`CREDIT_INCOME_RATIO`, `ANNUITY_INCOME_RATIO`)** show almost zero correlation with `TARGET`, but they may help in **non-linear models** (like tree-based algorithms).

```python
# Check missing data in percentage
missing_percent = df_prev.isnull().mean().sort_values(ascending=False) * 100
missing_percent = missing_percent.round(2)  # round to 2 decimals
missing_percent
```

<strong>Drop columns and handled missing values</strong>

```python
# Drop unnessaory and 40% of data is missing column from Application Data dataframe
missing_cols = [
    'RATE_INTEREST_PRIVILEGED',
    'RATE_INTEREST_PRIMARY',
    'AMT_DOWN_PAYMENT',
    'RATE_DOWN_PAYMENT',
    'NAME_TYPE_SUITE',
    'DAYS_TERMINATION',
    'DAYS_FIRST_DRAWING',
    'DAYS_FIRST_DUE',
    'DAYS_LAST_DUE_1ST_VERSION',
    'DAYS_LAST_DUE',
    'NFLAG_INSURED_ON_APPROVAL'
]

df_prev.drop(columns=missing_cols, inplace=True)
```

```python
# Fill missing with median, since they are numerical financial values.
num_cols = ['AMT_GOODS_PRICE', 'AMT_ANNUITY', 'CNT_PAYMENT', 'AMT_CREDIT']
df_prev[num_cols] = df_prev[num_cols].fillna(df_prev[num_cols].median())

# Fill missing with most frequent (mode).
df_prev['PRODUCT_COMBINATION'] = df_prev['PRODUCT_COMBINATION'].fillna(df_prev['PRODUCT_COMBINATION'].mode()[0])

# If output is 0, all missing values are handled successfully.
output = df_prev.isnull().sum().sum()
print(f"Missing Values Count : {output}")
```

```python
# Merge using inner
merged_df = pd.merge(df, df_prev, on='SK_ID_CURR', how='inner')
```

```python
# Calculate counts and percentages
status_counts = merged_df.groupby(['NAME_CONTRACT_STATUS', 'TARGET']).size().reset_index(name='count')
status_counts['percentage'] = status_counts.groupby('NAME_CONTRACT_STATUS')['count'].transform(lambda x: 100 * x / x.sum())

# Plot with explicit palette and correct legend
plt.figure(figsize=(8,5))
palette = {0: "skyblue", 1: "salmon"}  # Assign colors to TARGET values

sns.barplot(data=status_counts, x='NAME_CONTRACT_STATUS', y='percentage', hue='TARGET', palette=palette)
plt.title('Percentage of Loan Contract Status by Target (Default / Non-Default)')
plt.xlabel('Contract Status')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=45)

# Fix legend labels properly
handles, _ = plt.gca().get_legend_handles_labels()
plt.legend(handles=handles, title='TARGET', labels=['No Default (0)', 'Default (1)'])
plot_path = f"prev_loan_status.png"
plt.savefig(plot_path, bbox_inches='tight')
plt.show()
```

### Key Observations (Contract Status by Default Status)  - Previous Application Data

- **"Refused" Loans Have the Highest Default Rate:**
  - Among all the categories shown, the **"Refused"** contract status has the highest percentage of **Defaulters (1)**, at approximately 12-13%.
- **"Approved" and "Unused offer" Have the Lowest Default Rates:**
  - The default rate is lowest for clients whose previous contract was **"Approved"** or **"Unused offer,"** both sitting around 7-8%.
- **"Canceled" Status is Mid-Range:**
  - The default rate for clients with a **"Canceled"** previous contract is intermediate, around 9-10%.

```python
import matplotlib.pyplot as plt
import seaborn as sns

colsCategorical = [
    'TARGET','NAME_PAYMENT_TYPE','NAME_CLIENT_TYPE'
]

# Number of subplots
n_cols = 1
n_rows = 3

fig, axes = plt.subplots(n_rows, n_cols, figsize=(14, n_rows*5))
axes = axes.flatten()  # Flatten in case of multiple rows

for i, col in enumerate(colsCategorical):
    # Calculate percentage
    vc = merged_df_app[col].value_counts(normalize=True) * 100
    vc_df = vc.reset_index()
    vc_df.columns = [col, 'Percentage']
    
    # Barplot
    sns.barplot(
        data=vc_df,
        x=col,
        y='Percentage',
        hue=col,
        palette='viridis',
        dodge=False,
        ax=axes[i],
        legend=False
    )
    
    # Annotate bars
    for j, v in enumerate(vc_df['Percentage']):
        axes[i].text(j, v + 0.5, f"{v:.1f}%", ha='center', fontweight='bold')
    
    axes[i].set_title(f'{col} (% Distribution)', fontsize=12)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Percentage (%)')
    axes[i].tick_params(axis='x', rotation=45)

# Remove any empty subplots
for k in range(i+1, len(axes)):
    fig.delaxes(axes[k])

plt.tight_layout()
plot_path = f"prev_loan_status_2_{col}.png"
plt.savefig(plot_path, bbox_inches='tight')
plt.show()
```
